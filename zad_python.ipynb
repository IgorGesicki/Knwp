```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
df1 = pd.read_csv('kursy_eur_nbp_data.csv', sep=',', skiprows=1, header=None)
y2019=df1[df1[0]==2019][2].tolist()
y2020=df1[df1[0]==2020][2].tolist()
y2022=df1[df1[0]==2022][2].tolist()
test = df1[df1[0]==2019][1]
test = pd.to_datetime(test)
x = test.dt.dayofyear
x = x.tolist()

test2 = df1[df1[0]==2020][1]
test2 = pd.to_datetime(test2)
x2 = test2.dt.dayofyear
x2 = x2.tolist()

test3 = df1[df1[0]==2022][1]
test3 = pd.to_datetime(test3)
x3 = test3.dt.dayofyear
x3 = x3.tolist()

fig, ax = plt.subplots()
ax.plot(x, y2019, color = 'red', label='2019')
ax.plot(x2, y2020, color = 'blue', label='2020')
ax.plot(x3, y2022, color = 'blue', label='2022')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
df1 = pd.read_csv('kursy_chf_nbp_data.csv', sep=',', skiprows=1, header=None)

y2020=df1[df1[0]==2020][2].tolist()


test2 = df1[df1[0]==2020][1]
test2 = pd.to_datetime(test2)
x2 = test2.dt.dayofyear
x2 = x2.tolist()

avg=df1[df1[0]==2020][2].sum()/df1[df1[0]==2020].shape[0]

df_bool=df1[df1[0]==2020][2]<4
df_boolavg=df1[df1[0]==2020][2]<avg

fig, ax = plt.subplots()
days_under4=df_bool.sum()
days_underavg=df_boolavg.sum()
print("Kurs franka byl ponizej 4zl przez: ")
print(days_under4)
print("dni.")
print("Kurs franka byl ponizej sredniej przez: ")
print(days_underavg)
print("dni.")
ax.plot(x2, y2020, color = 'blue', label='2020')
plt.axhline(y=avg)
plt.legend()
plt.show()
```
